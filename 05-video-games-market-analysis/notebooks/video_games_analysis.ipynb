## Проверка ошибок в данных и их предобработка


### Проверка названий столбцов

На данном этапе проверяется корректность названий столбцов и выполняется приведение их к единому стилю snake_case.

```python
# Выводим на экран названия всех столбцов
df.columns
```

```python
# Приводим столбцы к стилю snake case, регистр - нижний, вместо пробелов - подчеркивания
df.columns = df.columns.str.lower().str.replace(' ', '_')

df.columns
```

### Типы данных

```python
# Выводим информацию о датасете для анализа типов данных
df.info()
```

### Вывод по результатам анализа типов данных

После изучения типов данных с помощью метода `info()` можно заметить, что некоторые столбцы имеют некорректные типы данных.

Например:

- столбцы `eu_sales` и `jp_sales` имеют тип object, хотя содержат числовые значения;
- столбец `user_score` также имеет тип object, хотя должен содержать числовые оценки пользователей.

Причиной появления типа `object` в числовых столбцах может быть наличие строковых значений в данных. Например, в столбце `user_score` могут встречаться текстовые значения, которые мешают корректному определению типа данных.

Для дальнейшего анализа необходимо привести эти столбцы к числовому типу данных.

```python
# Приводим столбец "Оценка пользователей" к числовому типу и заменяем строковые значения на пропуски
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce')
```

```python
# Приводим столбцы с числом продаж к числовому типу и заменяем строковые значения на пропуски
df['eu_sales'] = pd.to_numeric(df['eu_sales'], errors='coerce')
df['jp_sales'] = pd.to_numeric(df['jp_sales'], errors='coerce')
```

```python
# Выводим информацию о датасете для проверки результатов
df.info()
```

### Вывод по результатам анализа типов данных

В ходе проверки типов данных было выявлено несколько несоответствий.

Некоторые столбцы имели тип `object`, хотя должны содержать числовые значения. Вероятнее всего, из-за наличия строковых значений.

Столбцы `user_score`, `eu_sales` и `jp_sales` были преобразованы в числовой тип данных с помощью функции `pd.to_numeric()`. При преобразовании строковые значения были заменены на пропуски при помощи `errors='coerce'`.

После выполнения преобразований типы данных стали корректными, что позволит использовать их для дальнейшего анализа.

## Фильтрация данных

```python
# Формируем отдельный датафрейм с периодом выхода игр с 2000 по 2013 года (включительно)
df_actual = df[(df['year_of_release'] >= 2000) & (df['year_of_release'] <= 2013)].copy()
# Проверяем полученные результаты
df_actual.head()
```

### Вывод к отфильтрованным данным 

Для дальнейшего анализа был выбран период с 2000 по 2013 года включительно, так как именно этот период интересует заказчиков исследования.

Отфильтрованные данные сохранены в отдельный датафрейм `df_actual`.

## Категоризация данных
   

### Категоризация игр в зависимости от оценок пользователя <a class="tocSkip">

```python
def user_score_category(score):
    if score >= 8:
        return 'высокая оценка'
    elif score >= 3:
        return 'средняя оценка'
    else:
        return 'низкая оценка'
# Добавляем новый столбец
df_actual['user_score_category'] = df_actual['user_score'].apply(user_score_category)
```

### Категоризация игр в зависимости от оценок критиков <a class="tocSkip">

```python
def critic_score_category(score):
    if score >= 80:
        return 'высокая оценка'
    elif score >= 30:
        return 'средняя оценка'
    else:
        return 'низкая оценка'
# Добавляем новый столбец
df_actual['critic_score_category'] = df_actual['critic_score'].apply(critic_score_category)
```

### Группировка и подсчет игр по выделенным категориям оценки <a class="tocSkip">

```python
# Группируем и считаем количество игр по категориям оценок пользователей 
df_actual.groupby('user_score_category')['name'].count()
```

```python
# Группируем и считаем количество игр по категориям оценок критиков 
df_actual.groupby('critic_score_category')['name'].count()
```

### Топ-7 платформ по количеству выпущенных игр за актуальный период<a class="tocSkip">

```python
top_platforms = df_actual['platform'].value_counts().head(7)

top_platforms
```

### Результаты по категоризации игр 

После фильтрации данных за период 2000–2013 годов была проведена категоризация игр по оценкам пользователей и критиков.

Наибольшее количество игр относится к категории с низкими пользовательскими оценками. Это может говорить о том, что в рассматриваемый период выпускалось большое количество игр, которые не получали высоких оценок пользователей.

По оценкам критиков наблюдается преобладание игр со средними и низкими оценками. Количество игр с высокими оценками критиков значительно меньше, что подтверждает тот факт, что действительно высокооценённые игры встречаются реже.
