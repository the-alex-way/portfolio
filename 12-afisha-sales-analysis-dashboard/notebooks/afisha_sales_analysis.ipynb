## Описание данных

В исследовании используются три датасета.

Файл **final_tickets_orders_df.csv**

Содержит информацию о заказах билетов.
    
- `order_id` - уникальный идентификатор заказа 
- `user_id` - уникальный идентификатор пользователя 
- `created_dt_msk` - дата оформления заказа (московское время)
- `created_ts_msk` - дата и время оформления заказа (московское время)

- `cinema_circuit` - сеть кинотеатров; если неприменимо - значение `'нет'`

- `currency_code` - валюта оплаты (`rub`, `kzt`)

- `service_name` - билетный оператор 
- `tickets_count` - количество приобретённых билетов
- `total` - общая сумма заказа
- `days_since_prev` - количество дней с предыдущего заказа пользователя

Файл **final_tickets_events_df.csv**

Содержит информацию о мероприятиях.

- `event_id` - уникальный идентификатор мероприятия 
- `event_name` - название мероприятия 
- `event_type_description` - описание типа мероприятия 
- `event_type_main` - основная категория мероприятия 
- `organizers` = организатор мероприятия 
- `region_name` - регион проведения 
- `city_name` - город проведения 
- `city_id` - идентификатор города 
- `venue_id` - идентификатор площадки 
- `venue_name` - название площадки 
- `venue_address` - адрес площадки 

Файл **final_tickets_tenge_df.csv**

Содержит информацию о курсе казахстанского тенге к российскому рублю за 2024 год.

- `nominal` - номинал валюты (100 тенге) 
- `data` - дата курса 
- `curs` - курс 100 тенге к российскому рублю 
- `cdx` - обозначение валюты (`kzt`)

## Проверка корректности выгрузки данных

### Сопоставление идентификаторов мероприятий

- **Сравнение количества уникальных event_id**

```python
# Смотрим количество уникальных event_id в таблице заказов и справочнике
print(f"Уникальных мероприятий в таблице заказов (df_orders): {df_orders['event_id'].nunique()}")
print(f"Уникальных мероприятий в справочнике (df_events): {df_events['event_id'].nunique()}")
```

- **Поиск отсутствующих мероприятий**

```python
# Смотрим идентификаторы мероприятий, которые есть в заказах, но отсутствуют в справочнике
missing_events = (
    set(df_orders['event_id'])      # создаем множество уникальных event_id из заказов
    - set(df_events['event_id'])    # вычитаем все event_id из справочника
)

print(f'Количество отсутствующих мероприятий: {len(missing_events)}')
```

- **Подсчет количества заказов**

```python
# Отбираем все заказы, которые относятся к этим 19 мероприятиям
missing_orders = df_orders[df_orders['event_id'].isin(missing_events)]

print(f'Количество заказов: {len(missing_orders)}')
```

```python
# Смотрим первые строки 
missing_orders.head()
```

- **распределение проблемных заказов**

```python
# Проверяем, как распределены проблемные заказы по мероприятиям
missing_orders['event_id'].value_counts()
```

- **Проверка cinema_circuit**

```python
# Проверяем, являются ли проблемные заказы фильмами
missing_orders['cinema_circuit'].value_counts()
```

**Комментарий:**

Все 238 заказов имеют значение `cinema_circuit = 'нет'`. Следовательно, выявленное расхождение не связано с наличием сети кинотеатров в данных.

- **Расчет доли заказов с отсутствующими мероприятиями**

```python
# Доля заказов с отсутствующими мероприятиями
total_orders = len(df_orders)                                       # всего заказов
missing_orders_count = len(missing_orders)                          # заказов с отсутствующими мероприятиями
missing_percentage = (missing_orders_count / total_orders) * 100    # доля проблемных заказов от общего количества заказов

print(f'Всего заказов: {total_orders}')
print(f'Заказов с отсутствующими мероприятиями: {missing_orders_count}')
print(f'Доля заказов с отсутствующими мероприятиями: {missing_percentage:.2f}%')
```

**Комментарий:**

Для проверки корректности выгрузки были сопоставлены идентификаторы мероприятий в таблицах заказов и справочника мероприятий. Проверка показала, что в `таблице заказов содержится 22 446 уникальных мероприятий`, тогда как в `справочнике представлены 22 427 мероприятий`. Было обнаружено `19 идентификаторов мероприятий`, отсутствующих в справочнике, на которые приходится `238 заказов (0,08 % от общего числа заказов)`. 

Дополнительная проверка показала, что все эти заказы не относятся к сети кинотеатров (`cinema_circuit = 'нет'`), поэтому расхождение не связано с исключением фильмов при формировании выборки. 

Поскольку **доля таких заказов незначительна**, удалять их на данном этапе нецелесообразно. ## Создание новых признаков

### Выручка в рублях (revenue_rub)

```python
# Создаем столбец с выручкой в рублях (revenue_rub)
df_merged['revenue_rub'] = np.where(
    df_merged['currency_code'] == 'rub',
    df_merged['revenue'],                           # если рубли - оставляем как есть
    df_merged['revenue'] * df_merged['curs'] / 100  # если тенге - переводим в рубли
)

# Проверяем результат
df_merged[['currency_code', 'revenue', 'curs', 'revenue_rub']].head(10)
```

### Выручка с одного билета (one_ticket_revenue_rub) 

```python
# Создаем столбец с выручкой с продажи одного билета на мероприятие (one_ticket_revenue_rub)
df_merged['one_ticket_revenue_rub'] = df_merged['revenue_rub'] / df_merged['tickets_count']

# Проверяем
df_merged[['revenue_rub', 'tickets_count', 'one_ticket_revenue_rub']].head(10)
```

### Месяц заказа (month)

```python
# Создаем отдельный столбец с месяцем заказа (month)
df_merged['month'] = df_merged['created_dt_msk'].dt.month

# Проверяем 
df_merged[['created_dt_msk', 'month']].head()
```

### Сезон заказа (season)

```python
# Создаем столбец с информацией о сезонности (season)
def get_season(month):
    if month in [12, 1, 2]:
        return 'зима'
    elif month in [3, 4, 5]:
        return 'весна'
    elif month in [6, 7, 8]:
        return 'лето'
    elif month in [9, 10, 11]:
        return 'осень'

df_merged['season'] = df_merged['month'].apply(get_season)

# Проверяем
df_merged[['month', 'season']].head(10)
```

```python
# Проверяем количество заказов по
print('Распределение заказов по сезонам:')
print(df_merged['season'].value_counts())
```

Весна и зима отсутствуют, так как данные охватывают только период с июня по октябрь 2024 года.

### Подготовка осеннего датафрейма

```python
# Создаем отдельный датафрейм только с осенними заказами
df_autumn = (
    df_merged[df_merged['season'] == 'осень']
    .copy()
)
```

```python
# Проверяем результат
print(f'Количество заказов осенью: {len(df_autumn):,}')
print(
    f"Период: "
    f"{df_autumn['created_dt_msk'].min().date()} - "
    f"{df_autumn['created_dt_msk'].max().date()}"
)
```

**Комментарий:**

Создан осенний датафрейм. Он содержит 167 899 заказов за два месяца (сентябрь и октябрь 2024 года). Использование .copy() гарантирует независимость от исходного датафрейма. Период полностью соответствует заданию.

### Сводная таблица ежедневных метрик

```python
# Группируем данные по дням и рассчитываем основные показатели
daily_metrics = (
    df_autumn                                                  # берем только осенние заказы
    .groupby('created_dt_msk')                                 # группирует данные по дате заказа
    .agg(                                                      # применяем агрегацию к каждой группе
        orders_count=('order_id', 'count'),                    # количество заказов
        dau=('user_id', 'nunique'),                            # количество уникальных пользователей
        avg_ticket_price=('one_ticket_revenue_rub', 'mean')    # средняя стоимость одного билета
    )
    .reset_index()                                             # превращаем дату из индекса в обычный столбец
)

# Рассчитываем среднее число заказов на пользователя
daily_metrics['orders_per_user'] = (                           # создаем новый столбец в таблице
    daily_metrics['orders_count']                              # делим количество заказов 
    / daily_metrics['dau']                                     # на количество пользователей
)

# Округляем показатели для удобства отображения
daily_metrics['avg_ticket_price'] = (
    daily_metrics['avg_ticket_price'].round(2)
)

daily_metrics['orders_per_user'] = (
    daily_metrics['orders_per_user'].round(2)
)

# Проверяем результат
display(daily_metrics.head())
```

# Проверка статистических гипотез

## Гипотеза №1: Пользователи мобильных устройств оформляют больше заказов

- Проверим предположение, что пользователи мобильного приложения оформляют больше заказов, чем пользователи стационарных устройств.

**Нулевая гипотеза (H₀):**

- Среднее количество заказов на одного пользователя мобильных устройств **не больше**, чем у пользователей стационарных устройств.

**Альтернативная гипотеза (H₁):**

- Среднее количество заказов на одного пользователя мобильных устройств **больше**, чем у пользователей стационарных устройств.

### Группировка данных по пользователям и агрегация заказов

```python
# Рассчитываем количество заказов для каждого пользователя
user_orders = (
    df_autumn
    .groupby(['user_id', 'device_type_canonical'])
    .agg(
        orders_count=('order_id', 'count')
    )
    .reset_index()
)

# Проверяем результат
display(user_orders.head())
```

### Описательная статистика и обоснование выбора теста

```python
# Разделяем пользователей по типу устройства
mobile_orders = user_orders[
    user_orders['device_type_canonical'] == 'mobile'
]['orders_count']

desktop_orders = user_orders[
    user_orders['device_type_canonical'] == 'desktop'
]['orders_count']

# Выводим основные характеристики
print('Мобильные устройства')
display(mobile_orders.describe())

print('Стационарные устройства')
display(desktop_orders.describe())
```

**Выбор статистического теста**

Перед проверкой гипотезы были изучены описательные статистики для обеих выборок.

**Мобильные устройства (14 110 пользователей):**
- Среднее количество заказов: **9.47**
- Медиана: **2.00**
- Стандартное отклонение: **67.96**
- Максимальное значение: **4 314** заказов

**Стационарные устройства (4 863 пользователя):**
- Среднее количество заказов: **7.04**
- Медиана: **2.00**
- Стандартное отклонение: **35.90**
- Максимальное значение: **1 494** заказа

**Наблюдения:**
- В обоих группах среднее значение значительно превышает медиану (9.47 > 2.00 и 7.04 > 2.00), что указывает на **правостороннюю асимметрию** распределения.
- Стандартное отклонение существенно превышает среднее значение (67.96 > 9.47 и 35.90 > 7.04), что свидетельствует о **высокой вариативности** данных.
- Максимальные значения (4 314 и 1 494) значительно отличаются от типичных значений (большинство пользователей совершают 1-5 заказов), что подтверждает наличие **выбросов**.

Таким образом, распределение имеет выраженную асимметрию и содержит значительное количество выбросов, поэтому использование параметрического критерия может быть некорректным. Для сравнения двух независимых выборок выбран **непараметрический критерий Манна–Уитни**, который не требует выполнения предположения о нормальности распределения и устойчив к выбросам.

### Запуск теста и вывод

```python
# Задаем уровень значимости 5% (порог, с которым мы сравниваем p-value)
alpha = 0.05

stat, p_value = mannwhitneyu(    # проводим тест Манна–Уитни
    mobile_orders,               # первая выборка (пользователи mobile)
    desktop_orders,              # вторая выборка (пользователи desktop)
    alternative='greater'        # проверяем одностороннюю гипотезу (mobile > desktop)
)

print(f'Статистика критерия: {stat:.0f}')    # число, которое показывает, насколько сильно различаются группы
print(f'p-value: {p_value:.7f}')             # вероятность того, что наблюдаемая разница возникла случайно

if p_value < alpha:                                    # если p-value < 0.05, то результат статистически значимый
    print('Отвергаем нулевую гипотезу')                # в таком случае - отвергаем H₀
else:                                                  # если p-value ≥ 0.05, то нет оснований считать разницу значимой
    print('Не удалось отвергнуть нулевую гипотезу')    # в таком случае - не отвергаем H₀
```

**Вывод по гипотезе 1**

- По результатам критерия Манна-Уитни получено значение p-value = 0.0000002, что значительно меньше уровня значимости α = 0.05.
Следовательно, **нулевая гипотеза отвергается**.


- Полученные результаты позволяют сделать вывод, что **среднее количество заказов на одного пользователя мобильных устройств статистически значимо выше, чем у пользователей стационарных устройств**.

**Таким образом**, `альтернативная гипотеза нашла статистическое подтверждение`.

## Гипотеза №2: Время между заказами у пользователей мобильных устройств выше

- Проверим предположение, что среднее время между заказами пользователей мобильных устройств выше, чем у пользователей стационарных устройств.

**Нулевая гипотеза (H₀):**

- Среднее время между заказами пользователей мобильных устройств **не больше**, чем у пользователей стационарных устройств.

**Альтернативная гипотеза (H₁):**

- Среднее время между заказами пользователей мобильных устройств **больше**, чем у пользователей стационарных устройств.

### Расчет интервалов между заказами

```python
# Сортируем заказы по времени для каждого пользователя
df_sorted = df_autumn.sort_values(['user_id', 'created_ts_msk']).copy()
```

```python
# Считаем интервал между соседними заказами (в часах)
df_sorted['time_between'] = (
    df_sorted.groupby('user_id')['created_ts_msk']
    .diff()
    .dt.total_seconds() / 3600
)
```

```python
# Удаляем первые заказы (где нет предыдущего)
df_intervals = df_sorted.dropna(subset=['time_between'])
```

### Агрегация среднего времени на пользователя

```python
# Для каждого пользователя считаем среднее время между заказами
avg_time_per_user = (
    df_intervals
    .groupby(['user_id', 'device_type_canonical'])
    .agg(avg_time_between=('time_between', 'mean'))
    .reset_index()
)
```

```python
# Проверяем результат
print('Среднее время между заказами на пользователя:')
display(avg_time_per_user.head(10))
```

### Описательная статистика и обоснование выбора теста

```python
# Разделяем по типу устройства
mobile_avg_time = avg_time_per_user[avg_time_per_user['device_type_canonical'] == 'mobile']['avg_time_between']
desktop_avg_time = avg_time_per_user[avg_time_per_user['device_type_canonical'] == 'desktop']['avg_time_between']
```

```python
# Описательная статистика
print('Мобильные устройства (среднее время на пользователя):')
display(mobile_avg_time.describe())

print('Стационарные устройства (среднее время на пользователя):')
display(desktop_avg_time.describe())
```

**Выбор статистического теста**

Перед проверкой гипотезы были изучены описательные статистики для обеих выборок.

**Мобильные устройства (8 516 пользователей):**
- Среднее время между заказами: **150.70** часов
- Медиана: **74.52** часов
- Стандартное отклонение: **208.55** часов
- Максимальное значение: **1 408.42** часов

**Стационарные устройства (3 488 пользователей):**
- Среднее время между заказами: **127.78** часов
- Медиана: **47.93** часов
- Стандартное отклонение: **202.70** часов
- Максимальное значение: **1 408.55** часов

**Наблюдения:**
- В обеих группах среднее значение значительно превышает медиану (150.70 > 74.52 и 127.78 > 47.93), что указывает на **правостороннюю асимметрию** распределения.
- Стандартное отклонение существенно превышает среднее значение (208.55 > 150.70 и 202.70 > 127.78), что свидетельствует о **высокой вариативности** данных.
- Максимальные значения (1 408.42 и 1 408.55) значительно отличаются от типичных значений, что подтверждает наличие **выбросов**.

Таким образом, распределения не являются нормальными и содержат выбросы. Для сравнения используем **непараметрический критерий Манна-Уитни**, который не требует нормальности распределения и устойчив к выбросам.

### Запуск теста и вывод

```python
# Проверяем гипотезу
alpha = 0.05
stat, p_value = mannwhitneyu(
    mobile_avg_time,
    desktop_avg_time,
    alternative='greater'
)

print(f'U-statistic: {stat:.0f}')
print(f'p-value: {p_value:.11f}')

if p_value < alpha:
    print('Отвергаем H₀: время между заказами у mobile больше, чем у desktop')
else:
    print('Не удалось отвергнуть H₀')
```

**Вывод по гипотезе 2**

**Проверяемая гипотеза:** среднее время между заказами у пользователей мобильных устройств выше, чем у пользователей стационарных устройств.

**Результаты теста Манна–Уитни:**
- U-статистика: **15 973 954**
- p-value: **0.00000000004**

p-value значительно меньше 0.05, что позволяет **отвергнуть нулевую гипотезу**.

**Вывод:** альтернативная гипотеза (H₁) **подтверждена**. Среднее время между заказами у пользователей мобильных устройств статистически значимо выше, чем у пользователей стационарных устройств.

### Основные результаты анализа

**Сезонность и популярность мероприятий:**
- Количество заказов выросло в 3 раза с июня по октябрь (с 32 904 до 99 007 заказов).
- Осенью изменилась структура спроса:
  - Концерты теряют долю (-6.5 п.п.);
  - Театр и спорт растут (+4.6 и +8.8 п.п. соответственно);
  - Растут мероприятия с возрастным рейтингом 0+ (+5.27 п.п.), снижается доля 18+ (-3.43 п.п.).

**Средняя стоимость билета:**
- Снизилась у большинства категорий: театр (-18.6%), ёлки (-15.3%), концерты (-13.3%).
- Выросла только у выставок (+6.0%) и стендапа (+5.8%), что может свидетельствовать о росте интереса к этим форматам.

---

**Пользовательская активность:**


- За осень зафиксировано 167 899 заказов, DAU вырос к концу октября до 1 224 пользователей.
- Выявлены две аномальные даты - 3 сентября и 1 октября - с резким ростом числа заказов и падением средней цены (вероятно, акции).
- Недельная цикличность: в будни заказов на 22% больше, чем в выходные, но в выходные билеты на 10% дороже.

---

**Регионы и билетные партнёры**


- Явный лидер среди регионов - Каменевский регион: 26% мероприятий, 30% заказов.
- Второй по значимости - Североярская область: 17% мероприятий, 15% заказов.
- Билетные партнёры: «Билеты без проблем» - абсолютный лидер (20.5% заказов, 16.3% выручки).
- Топ-5 партнёров обеспечивают ~70% всех заказов.

---

**Проверка гипотез**


- **Гипотеза 1** (mobile > desktop по числу заказов): подтверждена. Пользователи mobile в среднем делают больше заказов.
- **Гипотеза 2** (mobile > desktop по времени между заказами): подтверждена. Пользователи mobile ждут между заказами дольше.

### Рекомендации:


- Усилить продвижение театра и спорта осенью - эти категории показывают рост спроса.
- Адаптировать маркетинг под семейную аудиторию - растёт доля мероприятий с рейтингом 0+.
- Сосредоточить рекламные бюджеты на Каменевском регионе - он приносит 30% всех заказов.
- Укрепить сотрудничество с топ-партнёрами («Билеты без проблем», «Лови билет!», «Билеты в руки») - они формируют основную выручку.
- Использовать аномалии (акции) для стимулирования спроса - даты 3 сентября и 1 октября показали высокую эффективность.
- Учитывать недельную цикличность - рекламные кампании эффективнее запускать в будни, а в выходные делать акцент на премиальные события.
