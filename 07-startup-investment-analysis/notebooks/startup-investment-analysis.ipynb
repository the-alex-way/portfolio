## Описание проекта

Финансовая компания, работающая на рынке венчурных инвестиций, планирует определить наиболее перспективные направления для инвестирования в стартапы. # Предобработка данных

План первоочередных действий по предобработке:

- Привести названия всех столбцов к единому стилю (нижний регистр, замена пробелов на подчёркивания).
- Очистить столбец `funding_total_usd` от запятых и преобразовать его в числовой тип.
- Преобразовать столбцы с датами (`*_at`) в формат `datetime`.
- Удалить явные дубликаты строк.
- Удалить строки, в которых отсутствует ключевая информация о финансировании (`funding_total_usd` is `NaN`).
- Разработать стратегию для заполнения пропусков в `mid_funding_at`

#### Приводим названия всех столбцов к единому стилю

```python
# Убираем пробелы и приводим к нижнему регистру ('snake_case')
df_inv.columns = df_inv.columns.str.strip().str.lower()
df_ret.columns = df_ret.columns.str.strip().str.lower()
```

```python
# Проверка преобразования названий столбцов
df_inv.columns
```

```python
# Проверка преобразования названий столбцов
df_ret.columns
```

#### Очищаем столбец `funding_total_usd` и приведим его к числовому типу

```python
# Выбирем столбец 'funding_total_usd' (объём привлечённых инвестиций)
# Преобразуем значение в строку и удаляем запятые
df_inv['funding_total_usd'] = (df_inv['funding_total_usd'].astype(str).str.replace(',', ''))

# Преобразуем строку в число
# В том случае, если что-то нельзя преобразовать в число, ставит NaN (пропуск) вместо ошибки
df_inv['funding_total_usd'] = pd.to_numeric(df_inv['funding_total_usd'], errors='coerce')
```

```python
# Выводим результат преобразования типа данных
df_inv['funding_total_usd'].dtype
```

#### Преобразуем типы данных в столбцах, которые хранят значения даты и времени

```python
# Создаем список с названиями столбцов, которые нужно преобразовать
date_cols = [
    'founded_at',          # дата основания компании
    'first_funding_at',    # дата первого финансирования
    'mid_funding_at',      # дата среднего по времени раунда финансирования
    'last_funding_at'      # дата последнего финансирования
]

# Создаем цикл для поочередного преобразования каждого элемента списка
# Преобразуем значения в дату
# В случае, если значение нельзя преобразовать в дату, ставит NaN (пропуск)
for col in date_cols:
    df_inv[col] = pd.to_datetime(df_inv[col], errors='coerce')
```

```python
# Выводим результат преобразования типа данных
df_inv[date_cols].dtypes
```

#### В датасете df_ret столбец year преобразуем в индекс

```python
# Делаем в качестве индекса датасета 'df_ret' столбец 'year'
df_ret = df_ret.set_index('year')
# Выводим результат
df_ret.head()
```

#### Обрабатываем текстовые данные

```python
# Создает список с текстовыми столбцами, в которых необходимо заполнить пропуски 
text_cols = ['name', 'homepage_url', 'category_list', 'market', 'status',
             'country_code', 'state_code', 'region', 'city']
# Заменяем одинаковым значением все пропуски
df_inv[text_cols] = df_inv[text_cols].fillna('unknown')
```

```python
# Проверяем результаты замены пропуском в текстовых столбцах
df_inv[text_cols].isna().sum()
```

#### Убираем полные дубликаты и пропуски в данных 

```python
# Удаляем полные дубликаты
df_inv = df_inv.drop_duplicates()
# Проверяем результат удаления
df_inv.duplicated().sum()
```

```python
# Удаляем строки, которые не несут какой-либо информации либо не содержат данных о финансировании в funding_total_usd.
df_inv = df_inv.dropna(subset=['funding_total_usd'])
# Проверяем результат удаления
df_inv['funding_total_usd'].isna().sum()
```

#### Заполняем пропуски в значениях mid_funding_at (используем середину интервала между first_funding_at и last_funding_at)

```python
# Заполняем пропуски
mask = df_inv['mid_funding_at'].isna()

# В качестве значения берем среднюю дату между первым и последним раундом
df_inv.loc[mask, 'mid_funding_at'] = (
    df_inv.loc[mask, 'first_funding_at'] +
    (df_inv.loc[mask, 'last_funding_at'] - df_inv.loc[mask, 'first_funding_at']) / 2
)
```

#### Разделим компании на группы по срокам финансирования
- Единичное финансирование (был всего один раунд финансирования)
- Срок финансирования до года (между первым и последним раундом финансирования прошло не более года)
- Срок финансирования более года

```python
# Создаем копию датасета
df = df_inv.copy()
```

```python
# Создаем новый столбец
# Вычисляем длительность финансирования в днях (дата последнего финансирования - дата первого финансирования)
df['funding_duration'] = (df['last_funding_at'] - df['first_funding_at']).dt.days
```

```python
# Создаем категориальный признак
# Определяем новую функцию 
def funding_group(row):
    
    if row['funding_rounds'] == 1:          # если у компании был только 1 раунд финансирования, то 
        return 'single_round'               # единичное финансирование
    elif row['funding_duration'] <= 365:    # если длительность финансирования не больше 365 дней (1 год), то
        return 'less_than_year'             # срок до года
    else:                                   # во всех остальных случаях (длительность больше 365 дней), то
        return 'more_than_year'             # срок более года
    
# Cохраняем результаты работы функции в новый столбец
# Функцию применяем горизонально к каждой строке 
df['funding_group'] = df.apply(funding_group, axis=1)
```

#### Выделяем сегменты рынка

- Массовый (более 120 компаний)
- Средний (от 35 до 120 включительно)
- Нишевый (до 35 компаний)

```python
# Считаем количество встречающихся значений в столбце 'market' (отрасль компании)
market_counts = df['market'].value_counts()

# Вводим функцию для классификации 
def get_group(x): 
    if x > 120:          # если в сегменте больше 120 компаний, то
        return 'mass'    # этот сегмент «массовый»
    elif x >= 35:        # если в сегменте от 35 до 120 компаний (включительно), то
        return 'mid'     # сегмент «средний» 
    else:                # если в сегменте меньше 35 компаний, то 
        return 'niche'   # сегмент «нишевый» 

# Вызываем функцию применительно к каждому значению
market_group = market_counts.apply(get_group)
```

```python
# Вывод количества сегментов в каждой категории
print(f"Массовых сегментов (> 120 компаний): {(market_group == 'mass').sum()}")
print(f"Средних сегментов (35-120): {(market_group == 'mid').sum()}")
print(f"Нишевых сегментов (< 35): {(market_group == 'niche').sum()}")
```

```python
# Замена значений в столбце 'market'
# Определяем новую функцию 
def replace_market(segment):                
    if segment in market_group.index:       # проверяем, есть ли название сегмента в индексе
        category = market_group[segment]
        if category == 'mass':              # если сегмент массовый, то
            return segment                  # возвращаем его исходное название
        elif category == 'mid':             # если сегмент средний, то
            return 'mid'                    # возвращаем строку 'mid' (заглушка)
        else:                               # если сегмент нишевый, то
            return 'niche'                  # возвращаем строку 'niche' (заглушка)
    return segment                          # если сегмента не было в market_group.index - возвращаем без изменений

# Применяем функцию к каждому значению в столбце
df['market'] = df['market'].apply(replace_market)
```

```python
# Создаем отдельный столбец для категорий 
# Если x не равен 'mid' и не равен 'niche', то вернуть 'mass'
df['market_group'] = df['market'].map(lambda x: 'mass' if x not in ['mid', 'niche'] else x)
```

```python
# Поиск выбросов (IQR по сегментам)
# Определяем функцию 
def detect_outliers(group):
    Q1 = group['funding_total_usd'].quantile(0.25) # 1-й квартиль для этой группы
    Q3 = group['funding_total_usd'].quantile(0.75) # 3-й квартиль для этой группы
    IQR = Q3 - Q1                                  # межквартильный размах
    
    lower = Q1 - 1.5 * IQR                         # нижняя граница для выбросов
    upper = Q3 + 1.5 * IQR                         # верхняя граница для выбросов
    
    # Определяем границы выбросов
    return (group['funding_total_usd'] < lower) | (group['funding_total_usd'] > upper)
```

```python
# Группируем строки по категории, выбираем только столбец с суммами, Применяет функцию к каждой группе
df['is_outlier'] = df.groupby('market_group')['funding_total_usd'] \
                     .transform(lambda x: detect_outliers(pd.DataFrame({'funding_total_usd': x})))
```

#### Годовая динамика доли возвращённых средств по типам финансирования

```python
# Годовая динамика доли возвращённых средств
# Считаем сумму инвестиций по годам и типам финансирования
# Группируем по году и суммируем инвестиции по каждому типу
investments_by_year = df_final.groupby('year')[funding_types].sum()

# Выводим таблицу (строки - годы, столбцы — типы финансирования, значения — суммарные инвестиции в этом году по каждому типу)
investments_by_year
```

```python
# Выбирает из датасета возвратов только те же типы финансирования
# Создает независимую копию 
returns_by_year = df_ret[funding_types].copy()
```

```python
# Рассчитываем долю возврата
# Добавляем маленькое число к знаменателю, чтобы избежать деления на ноль
epsilon = 1e-60

# Делим возвраты на инвестиции
# Переводим инвестиции из USD в миллионы USD
return_rate = returns_by_year / (investments_by_year / 1e6 + epsilon)  
```

```python
# Удаляем аномалии через IQR 
# Определяем новую функцию
def cap_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    # Возвращает очищенный от выбросов столбец
    return series.where((series >= lower) & (series <= upper), np.nan)

return_rate = return_rate.apply(cap_outliers_iqr)
```

```python
# Создает список из 5 типов финансирования, которые нужно отобразить на графике.
selected_types = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']
```

```python
# Строим график
plt.figure(figsize=(12, 6))

# Запускаем цикл по 5 выбранным типам финансирования
# Проверяем, есть ли такой столбец в данных 
for funding_type in selected_types:
    if funding_type in return_rate.columns:
        # Рисует линию на графике (годы - ось X, доля возврата для данного типа - ось Y)
        plt.plot(return_rate.index, return_rate[funding_type], 
                 marker='o', label=funding_type)

plt.title('Динамика доли возврата средств по типам финансирования')
plt.xlabel('Год')
plt.ylabel('Доля возврата (возвраты / инвестиции)')
plt.legend()
plt.grid()
plt.show()
```

```python
# Вывод статистики для анализа устойчивости
print("\nСредняя доля возврата по типам финандирования (за 2000-2014):")

# Вычисляет среднее значение по каждому столбцу, сортирует от большего к меньшему
avg_return = return_rate.mean().sort_values(ascending=False)

# Запускаем цикл по 5 типам 
for t in selected_types:
    if t in avg_return.index:
        # Выводит тип и среднюю долю с тремя знаками после запятой
        print(f"{t}: {avg_return[t]:.3f}")
```

Для удаления аномальных значений (выбросов) использовался метод IQR с коэффициентом 1.5. Значения, выходящие за границы `Q1 — 1.5×IQR` и `Q3 + 1.5×IQR`, были заменены на пропуски.

- **debt_financing** — стабильно высокая доля возврата (84.5%), линия на графике практически без разрывов. Это соответствует природе долгового финансирования: займы должны быть возвращены по договору.

- **angel** и **seed** — доли возврата ниже (65% и 52% соответственно), что характерно для ранних стадий инвестирования, где риск выше, а часть компаний закрывается без возврата средств.

- **venture** — самая низкая средняя доля возврата (43.8%), что объясняется высокорисковой природой венчурных инвестиций: много стартапов терпят неудачу, а успешные единицы окупают портфель целиком.

**Устойчивость показателя**

Наиболее устойчивый рост доли возврата наблюдается у **debt_financing** и **venture**:
- Линии этих типов имеют наименьшее количество разрывов
- Отсутствуют резкие обвалы после пиковых значений
- Тренд постепенного роста после 2008 года

Наименее устойчивы **private_equity** (разрывы в данных, резкие скачки) и **seed** (низкие значения, высокая волатильность).

**Вывод**

При формировании инвестиционной стратегии на 2015 год стоит учитывать:
- **debt_financing** — наиболее предсказуемый тип с точки зрения возврата средств
- **venture** — несмотря на низкую среднюю долю возврата, показывает устойчивую динамику после кризиса 2008 года
- **private_equity** — потенциально самый доходный, но требует более тщательного анализа из-за неполноты данных

#### Рекомендации для инвестиционной стратегии

**Перспективные области для финансирования**

- Рекомендация: обратить пристальное внимание на следующие массовые сегменты рынка: **`Technology`, `SaaS`, `Internet`, `Video`**.

- **Обоснование:**
- Мы проанализировали 54 массовых сегмента и выделили **13**, которые показали рост суммарного объема финансирования в 2014 году по сравнению с 2013 годом.
- Среди них по абсолютному объему привлеченных средств (топ-5) лидируют `Technology`, `Video`, `SaaS` и `Internet`.
- Сегменты `Technology` и `SaaS` демонстрируют **наиболее уверенный и устойчивый рост**, что говорит об их зрелости. Сегмент `Video` показывает взрывной рост, что может указывать на высокую доходность, но и на повышенные риски.

**Предпочтительный тип финансирования**

Рекомендация: использовать **комбинированную стратегию**, отдавая приоритет **венчурным инвестициям (`venture`)** и **долговому финансированию (`debt_financing`)**.

Обоснование:

`Venture` (венчурные инвестиции):
- Абсолютный лидер по объему привлеченных средств (> **104 млрд USD**) и популярности (использован **>18 тыс. компаний**).
- Несмотря на среднюю долю возврата (43.8%), этот тип финансирования критически важен для масштабирования бизнеса в высокотехнологичных отраслях.
     
`Debt_financing` (долговое финансирование):
- Самый надежный инструмент с точки зрения возврата средств. Демонстрирует **наиболее устойчивый рост доли возврата** (в среднем **84.5%**) и стабильную динамику.
- Идеально подходит для компаний с предсказуемым денежным потоком в зрелых сегментах (например, `SaaS`, `Technology`).
    
`Seed` и `Angel`: 
- Доля возврата ниже (52.3% и 65.1% соответственно), а риск выше, поэтому их использование в качестве основного инструмента не рекомендуется.

**Итоговая стратегия:**

Сфокусироваться на венчурных сделках в растущих массовых сегментах (`Technology`, `SaaS`). Для снижения рисков и обеспечения стабильного возврата капитала часть портфеля направить на долговое финансирование компаний из этих же сегментов, особенно на поздних стадиях развития.

**Заключение:** 

Несмотря на выявленные ограничения, проведенный анализ дает четкое и обоснованное направление для инвестиционной стратегии. Рекомендуется сфокусироваться на венчурных и долговых инструментах в растущих технологических сегментах, диверсифицируя портфель для снижения рисков.
