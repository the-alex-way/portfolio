#### Выявление пользователей, присутствующих в двух городах одновременно

```python
# Смотрим сколько пользователей в одном (двух) городах
# Группируем все строки по пользователю
# Cчитаем количество уникальных городов в записях пользователя
df.groupby('puid')['city'].nunique().value_counts()
```

**Комментарий:** обнаружены пользователи, которые одновременно присутствуют в обеих группах (Москва и Санкт-Петербург). Это нарушает независимость выборок, необходимую для корректного применения t-теста. Поэтому пользователей, имеющих более одного города в данных, необходимо исключить из анализа.

#### Исключение противоречивых пользователей из анализа

```python
# Оставляем только пользователей с непротиворечивыми данными (один город)
valid_users = df.groupby('puid')['city'].nunique()

# Оставляет в серии только тех пользователей, у которых один город, извлекает индексы ('puid')
valid_users = valid_users[valid_users == 1].index

# Проверяем для каждой строки находится ли 'puid' этой строки в списке непротиворчивых пользователей
df_clean = df[df['puid'].isin(valid_users)]
```

#### Построение гистограмм распределения активности

#### Построение boxplot-диаграмм для визуализации выбросов

```python
# Строим "ящик с усами" (boxplot)
plt.figure(figsize=(8,6))

sns.boxplot(
    data=df_clean, 
    x='city', 
    y='hours')

# Заголовок и подписи осей
plt.title('Распределение активности по городам')
plt.xlabel('Город')
plt.ylabel('Часы активности')

# Выводим результат
plt.show()
```

**Комментарий:** диаграмма размаха подтверждает наличие большого количества выбросов в обеих группах. При этом медианы групп находятся примерно на одном уровне.

#### Расчёт дисперсий для обеих групп

```python
# Сравниваем дисперсии групп
print('Дисперсия Москва:', np.var(moscow, ddof=1))
print('Дисперсия Санкт-Петербург:', np.var(spb, ddof=1))

# Разница дисперсий
variance_ratio = np.var(spb, ddof=1) / np.var(moscow, ddof=1)
print('Отношение дисперсий СПб / Москва:', variance_ratio)
```

**Комментарий по результатам сравнения дисперсий**

Для дополнительной оценки разброса значений были рассчитаны выборочные дисперсии:

- Москва: 1363.50
- Санкт-Петербург: 1586.57

Дисперсия активности пользователей Санкт-Петербурга несколько выше, чем у пользователей Москвы. Отношение дисперсий составляет примерно 1.16, то есть разброс значений в Санкт-Петербурге примерно на 16% больше.


Для проверки гипотезы целесообразно использовать t-тест Уэлча, который не требует предположения о равенстве дисперсий и корректно работает при различном размере выборок. 

## Проверка статистической гипотезы

#### Расчёт разницы средних

```python
# Разница средних
mean_diff = spb.mean() - moscow.mean()
print(f'Разница средних (СПб - Мск): {mean_diff:.2f} часов')
```

**Комментарий:** среднее время активности пользователей Санкт-Петербурга оказалось выше среднего времени активности пользователей Москвы на 0.42 часа (около 25 минут). Несмотря на положительное различие в пользу Санкт-Петербурга, его величина является небольшой относительно общего разброса данных внутри групп. 

#### Обоснование выбора статистического теста (t-тест Уэлча)

Для проверки гипотезы выбран t-тест Уэлча.

**Причины выбора:**

- сравниваются две независимые выборки;
- исследуется различие средних значений;
- размеры выборок достаточно большие (5990 и 2306 пользователей);
- дисперсии групп различаются;
- t-тест Уэлча устойчив к неравенству дисперсий и размеров выборок.

#### Формулировка гипотез

- **Нулевая гипотеза H₀:** Средняя активность пользователей в часах в двух группах (Москва и Санкт-Петербург) не различается.

- **Альтернативная гипотеза H₁:** Средняя активность пользователей в Санкт-Петербурге больше, и это различие статистически значимо.

#### Проведение t-теста Уэлча

```python
# Уровень статистической значимости (порог для принятия решения)
alpha = 0.05

# Проводим t-тест Уэлча
results = stats.ttest_ind(
    spb,                     # первая выборка (часы активности Петербурга)
    moscow,                  # вторая выборка (часы активности Москвы)
    equal_var=False,         # не предполагаем равенство дисперсий
    alternative='greater'    # среднее первой группы (spb) больше среднего второй (moscow)
)

print('t-статистика:', results.statistic)    # значение t-статистики
print('p-value:', results.pvalue)            # вероятность получить такие данные, если нулевая гипотеза верна
```

**Комментарий:**
- t-статистика: 0.435 - положительное число, значит среднее Петербурга действительно немного выше среднего Москвы (как и предполагала альтернативная гипотеза)

- p-value: 0.332 - это 33.2% вероятность получить такие данные, если на самом деле различий нет

#### Принятие статистического решения

```python
# Принимаем статистическое решение
# Если p-value меньше 0.05, мы отвергнем нулевую гипотезу
if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не удалось отвергнуть нулевую гипотезу')
```

**Комментарий по результатам проверки гипотезы:**

p-value (0.332) превышает пороговое значение 0.05. Это означает, что мы **не можем отвергнуть нулевую гипотезу** о равенстве средних. Наблюдаемое различие в средних между Петербургом и Москвой статистически **не значимо**.

Статистических оснований утверждать, что пользователи из Санкт-Петербурга проводят в приложении больше времени, чем пользователи из Москвы, **не обнаружено**. Различия в активности между городами могут быть объяснены случайной вариацией в выборках.

## Оценка достаточности выборки

- #### Задаем параметры
- #### Расчёт размера эффекта (proportion_effectsize)
- #### Расчёт минимального размера группы (NormalIndPower)

```python
# Параметры 
alpha = 0.05                  # уровень значимости (вероятность ошибки первого рода, т.е. ложноположительный результат)
power = 0.80                  # статистическая мощность (вероятность обнаружить эффект, если он действительно существует)

baseline_conversion = 0.30    # текущая (базовая) конверсия
expected_conversion = 0.33    # ожидаемая конверсия

# Рассчитываем размер эффекта
effect_size = proportion_effectsize(           # вычисляем меру различия между двумя пропорциями
    baseline_conversion,
    expected_conversion
)

# Создаём объект для расчёта мощности
# Реализует расчет мощности для теста разности средних/пропорций с использованием нормального распределения
analysis = NormalIndPower()       

# Рассчитываем минимальный размер группы
required_sample_size = analysis.solve_power(   # вычисляем размер выборки
    effect_size=effect_size,                   # размер эффекта
    power=power,                               # желаемая мощность теста
    alpha=alpha,                               # уровень значимости
    ratio=1                                    # соотношение размеров групп
)

# Выводим минимально необходимый размер выборки для каждой группы
round(required_sample_size)
```

## Расчёт конверсии по группам

#### Присоединение информации о группе к событиям

```python
# Добавляем группу пользователя к событиям
test_events = test_events.merge(
    eu_test_clean[['user_id', 'group']],    # берем из датафрейма колонки user_id и group (A или B)
    on='user_id',                           # ключ для объединения - по идентификатору пользователя
    how='left'                              # сохраняем все строки из test_events
)

# Проверяем
test_events.head()
```

#### Фильтрация событий покупки

```python
# Берем колонку с группами
# Подсчитываем количество записей (событий) в каждой группе
test_events['group'].value_counts()
```

- #### Подсчёт уникальных покупателей в каждой группе
- #### Объединение с общим количеством пользователей в группах

```python
# Расчет покупателей
buyers = (
    test_events                           # датафрейм событий
    .query('event_name == "purchase"')    # фильтруем - оставляем только покупки
    .groupby('group')['user_id']          # группируем по группам (A и B), берем колонку с идентификаторами пользователей
    .nunique()                            # считаем уникальных пользователей в каждой группе
    .reset_index(name='buyers')           # превращаем индекс (group) обратно в колонку и называет новую колонку buyers
)

# Считаем общее количество пользователей
users = (
    eu_test_clean                         # датафрейм с очищенными участниками теста
    .groupby('group')['user_id']          # группируем по группам A и B, берем колонку с идентификаторами
    .nunique()                            # считаем уникальных пользователей в каждой группе
    .reset_index(name='total_users')      # превращает индекс в колонку, называет новую колонку total_users
)

# Объединяем
conversion_data = users.merge(            # датафрейм с количеством пользователей в каждой группе
    buyers,                               # датафрейм с количеством покупателей в каждой группе
    on='group',                           # ключ для объединения - колонка group
    how='left'                            # сохраняем всех пользователей, добавляем данные о покупателях 
)

# Выводим результат
conversion_data
```

**Комментарий:**

- Количество пользователей в каждой группе получено из очищенной таблицы участников теста `eu_test_clean`.
- Количество покупателей получено фильтрацией событий по типу `purchase` с последующим подсчетом уникальных пользователей (один пользователь учитывается один раз, даже если совершил несколько покупок).
- Группа B (новый интерфейс) демонстрирует большее абсолютное число покупателей, однако для корректного вывода необходимо рассчитать конверсию с учетом разницы в размерах групп.

#### Проверяем изменение конверсии 

```python
# Расчет конверсии
conversion_data['conversion'] = (
    conversion_data['buyers']           # колонка с количеством покупателей
    / conversion_data['total_users']    # колонка с общим количеством пользователей   
)

# Создаем новую колонку с результатом деления
conversion_data
```

**Комментарий:** 

Для оценки эффективности интерфейса была рассчитана конверсия пользователей в покупку в каждой группе. Конверсия определялась как отношение числа пользователей, совершивших покупку, к общему числу пользователей в группе. Конверсия в группе B (тестовая, новый интерфейс) составила 29.54%, что выше конверсии группы A (контрольная, старый интерфейс) - 27.81%.

**Наблюдаемый эффект:**

Разница в конверсии составляет **1.7 процентных пункта** (с 27.8% до 29.5%). Абсолютное число покупателей в тестовой группе B выше на 103 человека (1480 против 1377).

**Предварительный вывод:**

Тестовая группа B (новый интерфейс) демонстрирует более высокую конверсию по сравнению с контрольной группой A. Однако для окончательного заключения необходимо провести **статистическую проверку значимости** полученных различий.

## Статистическая проверка гипотезы

#### Формулировка статистических гипотез

***Нулевая гипотеза (H₀)***

- Конверсия пользователей в покупку в тестовой группе B (новый интерфейс) не превышает конверсию в контрольной группе A (старый интерфейс).

***Альтернативная гипотеза (H₁)***

- Конверсия пользователей в покупку в тестовой группе B выше, чем в контрольной группе A.

#### Настройка параметров, односторонний тест

- Уровень значимости: α = 0,05
- Тип теста: односторонний
- Статистический критерий: z-тест для пропорций

**Правило принятия решения**

- Если p-value < 0,05 - отвергаем H₀, принимаем H₁ (новый интерфейс эффективнее)
- Если p-value ≥ 0,05 - не отвергаем H₀ (статистически значимых различий нет)

#### Подготовка данных для z-теста

```python
# количество успехов (покупатели)
success = [
    conversion_data.loc[conversion_data['group']=='B', 'buyers'].values[0],
    conversion_data.loc[conversion_data['group']=='A', 'buyers'].values[0]
]

# количество наблюдений (все пользователи)
nobs = [
    conversion_data.loc[conversion_data['group']=='B', 'total_users'].values[0],
    conversion_data.loc[conversion_data['group']=='A', 'total_users'].values[0]
]
```

**Комментарий:** для проведения z-теста данные подготовлены в порядке [B, A], так как при односторонней альтернативе `larger` проверяется, что конверсия первой группы (B) превышает конверсию второй (A).

#### Проведение одностороннего z-теста для пропорций

```python
# z-test
z_stat, p_value = proportions_ztest(
    success,
    nobs,
    alternative='larger'   # используем односторонний тест
)

z_stat, p_value
```

**Комментарий:** для проверки гипотезы об увеличении конверсии в группе B по сравнению с A применён односторонний z-тест для сравнения пропорций. Полученные значения: z-статистика = 1.9069 (положительное значение указывает на то, что конверсия в группе B выше), p-value = 0.0283.

#### Интерпретация

```python
# Интерпретация результата
alpha = 0.05

if p_value < alpha:
    print("Есть статистически значимое увеличение конверсии в группе B")
else:
    print("Статистически значимого увеличения конверсии не выявлено")
```

**Комментарий:** поскольку p-value (0.0283) меньше уровня значимости (0.05) можно утверждать, что новый интерфейс статистически значимо увеличивает конверсию пользователей в покупку по сравнению со старым интерфейсом.


## Оценка практической значимости и бизнес-цели

#### Расчёт прироста конверсии

```python
# Расчет практической значимости (бизнес-цель)
A = conversion_data.loc[conversion_data['group']=='A', 'conversion'].values[0]
B = conversion_data.loc[conversion_data['group']=='B', 'conversion'].values[0]

# Абсолютная разница в конверсии (B - A)
uplift_abs = B - A

# Относительная разница ((B - A) / A), переводим в проценты (* 100)
uplift_pct = (B - A) / A * 100

# Смотрим увеличение конверсии минимум на 3 процентных пункта.
meets_business_target = uplift_abs >= 0.03

# Выводим результаты
print(f"Абсолютный прирост конверсии: {uplift_abs:.4f} ({uplift_abs*100:.2f} п.п.)")
print(f"Относительный прирост (uplift): {uplift_pct:.2f}%")
print(f"Целевой ориентир (≥3 п.п.): {meets_business_target}")
```
